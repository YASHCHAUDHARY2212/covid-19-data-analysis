```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Matplotlib and Seaborn style
plt.style.use('seaborn-v0_8-darkgrid' if 'seaborn-v0_8-darkgrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 10
print("Data science environment ready!")
```

## 2. Ingesting & Inspecting Processed Time-Series Data
We load the cleaned tidy dataset and the country-level severity summary.

```python
df_tidy = pd.read_csv('../data/covid_global_tidy.csv')
df_summary = pd.read_csv('../data/country_risk_summary.csv')

print(f"Total Tidy Observations: {len(df_tidy):,}")
print(f"Total Countries Analyzed: {len(df_summary)}")
df_summary.head(10)
```

## 3. Visualizing Top 10 High-Risk Countries by Total Confirmed Infections
We examine the absolute infection burden across the 10 most impacted countries.

```python
top10 = df_summary.head(10)
plt.figure(figsize=(12, 6))
barplot = sns.barplot(x='Total_Confirmed', y='Country', data=top10, palette='viridis')
plt.title('Top 10 High-Risk Countries by Total Confirmed COVID-19 Cases', fontsize=14, fontweight='bold')
plt.xlabel('Total Confirmed Cases')
plt.ylabel('Country')

for p in barplot.patches:
    val = p.get_width()
    barplot.annotate(f"{val/1e6:.1f}M", (val + 1e6, p.get_y() + p.get_height()/2),
                     va='center', fontsize=9, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 4. Visualizing Regional Spikes & Epidemic Waves (Resume Core Feature)
We plot daily new cases and 7-day moving averages for hotspot nations (India, US, Brazil, UK) to highlight epidemic wave peaks.

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10), sharex=False)
hotspots = [
    ('India', axes[0, 0], 'Delta Wave Peak (May 2021): ~414k/day'),
    ('US', axes[0, 1], 'Omicron Wave Peak (Jan 2022): ~1.3M/day'),
    ('Brazil', axes[1, 0], 'Multiple High-Mortality Surges'),
    ('United Kingdom', axes[1, 1], 'Alpha and Omicron Waves')
]

for country, ax, label in hotspots:
    c_df = df_tidy[df_tidy['Country'] == country].copy()
    c_df['Date'] = pd.to_datetime(c_df['Date'])
    ax.bar(c_df['Date'], c_df['Daily_Cases'], color='crimson', alpha=0.35, label='Daily Reported Cases')
    ax.plot(c_df['Date'], c_df['Daily_Cases_7d_MA'], color='deepskyblue', linewidth=2, label='7-Day Moving Avg')
    ax.set_title(f"{country}: Daily Infection Spikes & Wave Trends", fontweight='bold')
    ax.set_ylabel('Daily Cases')
    ax.legend(loc='upper left')

plt.suptitle('Regional COVID-19 Spikes & Wave Peak Detection (Matplotlib)', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 5. Case Fatality Rate (CFR %) Distribution & Healthcare Strain
Comparing the relationship between total caseload and mortality ratios to evaluate healthcare resilience.

```python
qualified = df_summary[df_summary['Total_Confirmed'] >= 100000]
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Total_Confirmed', y='CFR_Pct', hue='Risk_Category',
                data=qualified, palette={'High Risk Zone': 'red', 'Moderate Risk Zone': 'orange', 'Controlled / Low Risk Zone': 'green'}, s=100)
plt.xscale('log')
plt.title('Case Fatality Rate (CFR %) vs Confirmed Cases (Log Scale)', fontsize=14, fontweight='bold')
plt.xlabel('Total Confirmed Cases (Log Scale)')
plt.ylabel('Case Fatality Rate (%)')
plt.grid(True, which='both', ls='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 6. Key Conclusions & Findings
- **Wave Dynamics:** Omicron in late 2021/early 2022 produced unprecedented transmission spikes (US daily cases > 1.3M), but Delta (May 2021) exhibited significantly higher hospitalizations and mortality.
- **Reporting Smoothing:** 7-day rolling moving averages successfully removed weekend reporting artifacts.
- **High-Risk Zones:** Countries with high population density and healthcare bottlenecks (India, Brazil, Mexico, Peru) faced extreme peak strain requiring urgent intervention.
